# Finding 2B- 1
# The median spending of customers is approximately ₹6,415 for the

# 36–45 age group, ₹6,791 for the 46–55 age group, and ₹7,392 for
# the 56+ age group.

# These median values represent the middle spending level of
# customers within each age group.

# We want to understand how customer spending varies within each age group,
# rather than relying only on a single central value such as the median.

```python
df.groupby("age_group")["total_spend_inr"].agg(
    ["count", "min", "median", "mean", "max"]
)
```

### Finding 2B- 2

Customer spending varies considerably within every age group. The
minimum and maximum spending values show a wide range of individual
customer spending.

For example, in the 26–35 age group, spending ranges from approximately
₹472 to ₹54,268, while the median spending is ₹6,460 and the mean
spending is ₹8,195.

The mean is higher than the median across all age groups, indicating
that some higher-spending customers are pulling the average upward.

The 56+ group has the highest median and mean spending, but this group
contains only 8 customers, so the result should be interpreted with
caution.

Question 8: Which product category generates the most spending?

```python
df.groupby('product_category')['total_spend_inr'].sum().sort_values(ascending=False)
```

### Finding

Question 8.1 Which category has the highest spending per customer?

```python
df.groupby("product_category")["total_spend_inr"].mean().sort_values(ascending=False)
```

### Finding
Fashion generates the highest total spending overall, but Books has
the highest average spending per customer at approximately ₹8,380.
This shows that Fashion's high total spending is influenced by its
larger customer base, while individual customers in Books have the
highest average spending.

Why mean() here?
Because every row represents one customer, and total_spend_inr is that customer's total spending.
So the mean gives us:
Average spending per customer in each category.
